```python
# Automatic Quiz Generator

import re
import random

text = """
Artificial Intelligence is a branch of computer science that enables
machines to perform tasks that normally require human intelligence.
Machine learning is a subset of artificial intelligence that allows
computers to learn from data without being explicitly programmed.
Natural Language Processing enables computers to understand and
process human language.
"""

# Extract important words
words = re.findall(r'\b[A-Za-z]{5,}\b', text.lower())

stop_words = {
    "which", "their", "there", "these", "those", "about",
    "allows", "without", "being", "normally", "requires"
}

important_words = [
    word for word in words
    if word not in stop_words
]

word_frequency = {}

for word in important_words:
    word_frequency[word] = word_frequency.get(word, 0) + 1

# Select important keywords
keywords = sorted(
    word_frequency,
    key=word_frequency.get,
    reverse=True
)[:5]

print("=" * 55)
print("             AUTOMATIC QUIZ GENERATOR")
print("=" * 55)

print("\nSource Text:")
print(text.strip())

print("\nGenerated Questions:\n")

for i, keyword in enumerate(keywords, 1):

    question = (
        f"Q{i}. Which concept is related to "
        f"the following topic: '{keyword}'?"
    )

    # Create options
    options = [keyword]

    other_words = [
        w for w in important_words
        if w != keyword
    ]

    random.shuffle(other_words)

    for word in other_words:
        if word not in options:
            options.append(word)

        if len(options) == 4:
            break

    random.shuffle(options)

    print(question)

    for j, option in enumerate(options):
        print(f"   {chr(65+j)}) {option}")

    correct_option = chr(65 + options.index(keyword))

    print(f"   Answer: {correct_option}) {keyword}")
    print()

print("=" * 55)
print("Quiz Generated Successfully!")
print("=" * 55)
```
